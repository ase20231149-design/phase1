````python
from pathlib import Path
import pandas as pd
import numpy as np
import requests
import time
import re
import json

PROJECT_DIR = Path.cwd()

DATA_DIR = PROJECT_DIR / "data"
PDF_DIR = DATA_DIR / "pdfs"
TEXT_DIR = DATA_DIR / "texts"
MODELS_DIR = PROJECT_DIR / "models"
REPORTS_DIR = PROJECT_DIR / "reports"
AI_LOGS_DIR = PROJECT_DIR / "ai-logs" / "phase1"

for folder in [PDF_DIR, TEXT_DIR, MODELS_DIR, REPORTS_DIR, AI_LOGS_DIR]:
    folder.mkdir(parents=True, exist_ok=True)

print("Project location:", PROJECT_DIR)
````

````python

CATEGORY = "cs.CV"
START_DATE = "2023-01-01"
END_DATE = "2026-10-31"

SUBTOPICS = {
    "object_detection": "object detection",
    "image_segmentation": "image segmentation",
    "medical_imaging": "medical image analysis",
    "pose_estimation": "human pose estimation"
}

TARGET_PAPERS = 500
PAPERS_PER_TOPIC = 150
REQUEST_DELAY = 3.2

print("Category:", CATEGORY)
print("Date range:", START_DATE, "to", END_DATE)
print("Number of topics:", len(SUBTOPICS))
print("Target papers:", TARGET_PAPERS)
````

````python
import xml.etree.ElementTree as ET

API_URL = "https://export.arxiv.org/api/query"
all_papers = []
namespace = {
    "atom": "http://www.w3.org/2005/Atom"
}

for subtopic, keywords in SUBTOPICS.items():
    print("Searching:", keywords)

    query = (
        f'cat:{CATEGORY} AND all:"{keywords}" '
        f'AND submittedDate:[202301010000 TO 202610312359]'
    )

    parameters = {
        "search_query": query,
        "start": 0,
        "max_results": PAPERS_PER_TOPIC,
        "sortBy": "submittedDate",
        "sortOrder": "descending"
    }

    response = requests.get(API_URL, params=parameters, timeout=60)
    response.raise_for_status()

    root = ET.fromstring(response.content)
    entries = root.findall("atom:entry", namespace)

    for entry in entries:

        paper_id = entry.find("atom:id", namespace).text.strip()
        title = entry.find("atom:title", namespace).text.strip()
        abstract = entry.find("atom:summary", namespace).text.strip()
        published = entry.find("atom:published", namespace).text.strip()

        paper = {
            "title": " ".join(title.split()),
            "abstract": " ".join(abstract.split()),
            "arxiv_id": paper_id.split("/")[-1],
            "link": paper_id,
            "published": published,
            "subtopic": subtopic
        }

        all_papers.append(paper)

    print("Papers found:", len(entries))

    time.sleep(REQUEST_DELAY)


metadata_df = pd.DataFrame(all_papers)

metadata_df = metadata_df.drop_duplicates(
    subset="arxiv_id"
).reset_index(drop=True)

metadata_df.to_csv(
    DATA_DIR / "arxiv_metadata.csv",
    index=False,
    encoding="utf-8-sig"
)

print("\nCollection completed!")
print("Total unique papers:", len(metadata_df))

print("\nPapers per topic:")
print(metadata_df["subtopic"].value_counts())
````

````python

downloaded = 0
failed = 0

for index, paper in metadata_df.iterrows():

    arxiv_id = paper["arxiv_id"]

    pdf_url = f"https://arxiv.org/pdf/{arxiv_id}"

    pdf_path = PDF_DIR / f"{arxiv_id}.pdf"

    if pdf_path.exists():
        downloaded += 1
        continue

    try:
        response = requests.get(pdf_url, timeout=60)

        if response.status_code == 200 and response.content[:4] == b"%PDF":

            with open(pdf_path, "wb") as file:
                file.write(response.content)

            downloaded += 1

        else:
            failed += 1

    except Exception:
        failed += 1

    time.sleep(REQUEST_DELAY)

    if (index + 1) % 10 == 0:
        print("Processed:", index + 1, "of", len(metadata_df))

print("\nDownload completed!")
print("Downloaded:", downloaded)
print("Failed:", failed)
````

````python
import pymupdf

paper_data = []

for index, paper in metadata_df.iterrows():

    arxiv_id = paper["arxiv_id"]

    pdf_path = PDF_DIR / f"{arxiv_id}.pdf"
    text_path = TEXT_DIR / f"{arxiv_id}.txt"

    if not pdf_path.exists():
        continue

    try:
        pdf = pymupdf.open(pdf_path)

        full_text = ""

        for page in pdf:
            full_text += page.get_text()

        page_count = len(pdf)
        pdf.close()

        with open(text_path, "w", encoding="utf-8") as file:
            file.write(full_text)

        if page_count < 6:
            length_category = "short"
        elif 7 <= page_count <= 9:
            length_category = "medium"
        elif page_count > 9:
            length_category = "long"
        else:
            length_category = "unclassified"

        paper_data.append({
            "arxiv_id": arxiv_id,
            "page_count": page_count,
            "extracted_characters": len(full_text),
            "length_category": length_category,
            "extraction_status": "success"
        })

    except Exception as error:

        paper_data.append({
            "arxiv_id": arxiv_id,
            "page_count": 0,
            "extracted_characters": 0,
            "length_category": "unclassified",
            "extraction_status": "failed"
        })

    if (index + 1) % 25 == 0:
        print("Processed:", index + 1, "of", len(metadata_df))

extraction_df = pd.DataFrame(paper_data)

print("\nExtraction completed!")
print("Successful:", (extraction_df["extraction_status"] == "success").sum())
print("Failed:", (extraction_df["extraction_status"] == "failed").sum())
````

````python

corpus_df = pd.merge(
    metadata_df,
    extraction_df,
    on="arxiv_id",
    how="inner"
)

corpus_df.to_csv(
    DATA_DIR / "corpus.csv",
    index=False,
    encoding="utf-8-sig"
)

print("Total papers:", len(corpus_df))
print("\nPapers per topic:")
print(corpus_df["subtopic"].value_counts())
print("\nPaper length categories:")
print(corpus_df["length_category"].value_counts())
print("\nExtraction status:")
print(corpus_df["extraction_status"].value_counts())
print("\nDataset saved successfully!")
````

````python
documents = []
valid_rows = []

for index, paper in corpus_df.iterrows():

    arxiv_id = paper["arxiv_id"]
    text_path = TEXT_DIR / f"{arxiv_id}.txt"

    if text_path.exists():

        with open(text_path, "r", encoding="utf-8") as file:
            paper_text = file.read()

        if len(paper_text) > 100:

            document = (
                str(paper["title"]) + " " +
                str(paper["abstract"]) + " " +
                paper_text
            )

            documents.append(document)
            valid_rows.append(index)

corpus_df = corpus_df.loc[valid_rows].reset_index(drop=True)

corpus_df.to_csv(
    DATA_DIR / "corpus.csv",
    index=False,
    encoding="utf-8-sig"
)

print("Documents ready:", len(documents))
print("Corpus records:", len(corpus_df))
print("First document characters:", len(documents[0]))
````

````python
from textblob import TextBlob

abbreviations = {
    "ai": "artificial intelligence",
    "cv": "computer vision",
    "cnn": "convolutional neural network",
    "gan": "generative adversarial network",
    "vit": "vision transformer",
    "yolo": "you only look once"
}

def process_query(query):

    
    query = query.lower()

    words = query.split()
    expanded_words = []

    for word in words:
        if word in abbreviations:
            expanded_words.append(abbreviations[word])
        else:
            expanded_words.append(word)

    query = " ".join(expanded_words)

    
    corrected_query = str(TextBlob(query).correct())

    return corrected_query


test_query = "AI for objet detection"

processed_query = process_query(test_query)

print("Original query:", test_query)
print("Processed query:", processed_query)
````

````python


def hybrid_search(query, top_k=10):

    phrases = re.findall(r'"([^"]+)"', query)

    processed_query = process_query(query)

    query_tokens = re.findall(
        r"\b[a-zA-Z0-9]+\b",
        processed_query.lower()
    )


    bm25_scores = bm25_model.get_scores(query_tokens)
    bm25_ranking = np.argsort(bm25_scores)[::-1]


    query_embedding = embedding_model.encode(
        [processed_query],
        normalize_embeddings=True
    )

    dense_scores = np.dot(embeddings, query_embedding[0])
    dense_ranking = np.argsort(dense_scores)[::-1]


    rrf_scores = {}
    k = 60

    for rank, doc_index in enumerate(bm25_ranking):
        rrf_scores[doc_index] = (
            rrf_scores.get(doc_index, 0) +
            1 / (k + rank + 1)
        )

    for rank, doc_index in enumerate(dense_ranking):
        rrf_scores[doc_index] = (
            rrf_scores.get(doc_index, 0) +
            1 / (k + rank + 1)
        )


    final_ranking = sorted(
        rrf_scores,
        key=rrf_scores.get,
        reverse=True
    )

    results = []

    for doc_index in final_ranking[:top_k]:

        paper = corpus_df.iloc[doc_index]

        results.append({
            "title": paper["title"],
            "arxiv_id": paper["arxiv_id"],
            "link": paper["link"],
            "subtopic": paper["subtopic"],
            "score": rrf_scores[doc_index],
            "snippet": documents[doc_index][:200]
        })

    return pd.DataFrame(results)
````

````python
def search_papers(query, method="Hybrid", top_k=10):

    processed_query = process_query(query)

    query_tokens = re.findall(
        r"\b[a-zA-Z0-9]+\b",
        processed_query.lower()
    )

    if method in ["BM25", "Hybrid"]:
        bm25_scores = bm25_model.get_scores(query_tokens)

    if method in ["Dense", "Hybrid"]:
        query_embedding = embedding_model.encode(
            [processed_query],
            normalize_embeddings=True
        )

        dense_scores = np.dot(
            embeddings,
            query_embedding[0]
        )

    if method == "BM25":
        final_scores = bm25_scores

    elif method == "Dense":
        final_scores = dense_scores

    elif method == "Hybrid":

        bm25_ranking = np.argsort(bm25_scores)[::-1]
        dense_ranking = np.argsort(dense_scores)[::-1]

        
        rrf_scores = {}
        k = 60

        for rank, doc_index in enumerate(bm25_ranking):
            rrf_scores[doc_index] = (
                1 / (k + rank + 1)
            )

        for rank, doc_index in enumerate(dense_ranking):
            rrf_scores[doc_index] = (
                rrf_scores.get(doc_index, 0) +
                1 / (k + rank + 1)
            )

        final_scores = np.zeros(len(documents))

        for doc_index, score in rrf_scores.items():
            final_scores[doc_index] = score


    final_ranking = np.argsort(final_scores)[::-1]

    results = []


    for doc_index in final_ranking[:top_k]:

        paper = corpus_df.iloc[doc_index]

        results.append({
            "title": paper["title"],
            "arxiv_id": paper["arxiv_id"],
            "link": paper["link"],
            "subtopic": paper["subtopic"],
            "score": final_scores[doc_index],
            "snippet": documents[doc_index][:200]
        })

    return pd.DataFrame(results)
````

````python
app_code = r'''
import streamlit as st
import pandas as pd
import numpy as np
import pickle
import json
import re
from pathlib import Path
from sentence_transformers import SentenceTransformer
from textblob import TextBlob

# Project folders
PROJECT_DIR = Path(__file__).resolve().parent
DATA_DIR = PROJECT_DIR / "data"
MODELS_DIR = PROJECT_DIR / "models"

# Load saved data
corpus_df = pd.read_pickle(DATA_DIR / "corpus.pkl")

with open(DATA_DIR / "documents.json", "r", encoding="utf-8") as file:
    documents = json.load(file)

with open(MODELS_DIR / "bm25_model.pkl", "rb") as file:
    bm25_model = pickle.load(file)

embeddings = np.load(MODELS_DIR / "document_embeddings.npy")

# Load embedding model
@st.cache_resource
def load_embedding_model():
    return SentenceTransformer("all-MiniLM-L6-v2")

embedding_model = load_embedding_model()

# Abbreviations
abbreviations = {
    "ai": "artificial intelligence",
    "cv": "computer vision",
    "cnn": "convolutional neural network",
    "gan": "generative adversarial network",
    "vit": "vision transformer",
    "yolo": "you only look once"
}

def process_query(query):
    phrases = re.findall(r'"([^"]+)"', query)
    query = query.lower()
    words = query.split()
    expanded_words = []

    for word in words:
        if word in abbreviations:
            expanded_words.append(abbreviations[word])
        else:
            expanded_words.append(word)

    query = " ".join(expanded_words)
    query = str(TextBlob(query).correct())

    return query, phrases

def search_papers(query, method="Hybrid", top_k=10):
    processed_query, phrases = process_query(query)

    query_tokens = re.findall(
        r"\b[a-zA-Z0-9]+\b",
        processed_query.lower()
    )

    bm25_scores = bm25_model.get_scores(query_tokens)

    query_embedding = embedding_model.encode(
        [processed_query],
        normalize_embeddings=True
    )

    dense_scores = np.dot(embeddings, query_embedding[0])

    if method == "BM25":
        final_scores = bm25_scores

    elif method == "Dense":
        final_scores = dense_scores

    else:
        bm25_ranking = np.argsort(bm25_scores)[::-1]
        dense_ranking = np.argsort(dense_scores)[::-1]

        rrf_scores = {}
        k = 60

        for rank, doc_index in enumerate(bm25_ranking):
            rrf_scores[doc_index] = 1 / (k + rank + 1)

        for rank, doc_index in enumerate(dense_ranking):
            rrf_scores[doc_index] = (
                rrf_scores.get(doc_index, 0) +
                1 / (k + rank + 1)
            )

        final_scores = np.zeros(len(documents))

        for doc_index, score in rrf_scores.items():
            final_scores[doc_index] = score

    # Phrase matching boost
    for doc_index, document in enumerate(documents):
        for phrase in phrases:
            if phrase.lower() in document.lower():
                final_scores[doc_index] += 0.01

    final_ranking = np.argsort(final_scores)[::-1]

    results = []

    for doc_index in final_ranking[:top_k]:
        paper = corpus_df.iloc[doc_index]

        results.append({
            "title": paper["title"],
            "arxiv_id": paper["arxiv_id"],
            "link": paper["link"],
            "subtopic": paper["subtopic"],
            "score": round(float(final_scores[doc_index]), 5),
            "snippet": documents[doc_index][:200]
        })

    return pd.DataFrame(results)

# Streamlit interface
st.set_page_config(
    page_title="arXiv Hybrid Search",
    page_icon="📚",
    layout="wide"
)

st.title("arXiv Research Paper Search")
st.write("Search research papers using BM25, Dense Retrieval, or Hybrid Search.")

query = st.text_input("Enter your search query:")

method = st.selectbox(
    "Choose search method:",
    ["BM25", "Dense", "Hybrid"]
)

if st.button("Search"):
    if query.strip():

        with st.spinner("Searching papers..."):
            results = search_papers(query, method, 10)

        st.subheader("Top 10 Results")

        for index, paper in results.iterrows():
            st.markdown(f"### {index + 1}. {paper['title']}")
            st.write("Sub-topic:", paper["subtopic"])
            st.write("arXiv ID:", paper["arxiv_id"])
            st.write("Score:", paper["score"])
            st.write(paper["snippet"] + "...")
            st.markdown(f"[Open paper on arXiv]({paper['link']})")
            st.divider()

    else:
        st.warning("Please enter a search query.")
'''

app_path = PROJECT_DIR / "app.py"

with open(app_path, "w", encoding="utf-8") as file:
    file.write(app_code)

print("Streamlit app created successfully!")
print("App file exists:", app_path.exists())
print("App location:", app_path)
````
